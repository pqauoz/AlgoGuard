```python
import numpy as np
import os

# Define the path to your processed data
data_dir = "../dataset/processed"

# Load the train and test sets (These have 145 features!)
X_train = np.load(os.path.join(data_dir, "X_train.npy"))
y_train = np.load(os.path.join(data_dir, "y_train.npy"))
X_test = np.load(os.path.join(data_dir, "X_test.npy"))
y_test = np.load(os.path.join(data_dir, "y_test.npy"))

print("✓ Data loaded successfully!")
print(f"X_train shape: {X_train.shape}") # This should print (65914, 145)
print(f"y_train shape: {y_train.shape}")
```

```python
import os
import time
import joblib
import psutil
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_auc_score
)

os.makedirs("../models", exist_ok=True)
os.makedirs("../results", exist_ok=True)

models = {
    "Random_Forest": RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, random_state=42),
    "Gradient_Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5, n_jobs=-1),
    "Naive_Bayes": GaussianNB()
}

results = []

for name, model in models.items():
    print("=" * 70)
    print(f"Training {name}")

    process = psutil.Process(os.getpid())
    cpu_before = psutil.cpu_percent(interval=None)
    ram_before = process.memory_info().rss

    # Train
    start_train = time.time()
    model.fit(X_train, y_train)
    end_train = time.time()

    cpu_after = psutil.cpu_percent(interval=None)
    ram_after = process.memory_info().rss

    # Predict
    start_predict = time.time()
    predictions = model.predict(X_test)
    end_predict = time.time()

    # Core Metrics
    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions)
    recall = recall_score(y_test, predictions)
    f1 = f1_score(y_test, predictions)

    # Fixed Confusion Matrix
    tn, fp, fn, tp = confusion_matrix(y_test, predictions, labels=[0, 1]).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0

    # ROC-AUC
    if hasattr(model, "predict_proba"):
        try:
            probabilities = model.predict_proba(X_test)[:, 1]
            roc = roc_auc_score(y_test, probabilities)
        except Exception:
            roc = np.nan
    else:
        roc = np.nan

    # Save Model
    filename = f"../models/{name}.pkl"
    joblib.dump(model, filename)
    model_size = os.path.getsize(filename) / (1024 * 1024)

    cpu_usage = (cpu_before + cpu_after) / 2

    results.append({
        "Model": name,
        "Features": X_train.shape[1], 
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "False Positive Rate": fpr,
        "ROC-AUC": roc,
        "CPU Usage (%)": cpu_usage,
        "RAM Usage (MB)": abs((ram_after - ram_before) / (1024 * 1024)),
        "Training Time (s)": end_train - start_train,
        "Prediction Time (s)": end_predict - start_predict,
        "Model Size (MB)": model_size
    })

    print(f"✓ {name} completed. | Specificity: {specificity:.4f} | FPR: {fpr:.4f} | Sum: {specificity + fpr:.4f}")

results_df = pd.DataFrame(results)
display(results_df)

results_df.to_csv("../results/base_models_results.csv", index=False)
print("\nAll base models trained successfully on 145 features!")
```

```python
import time
import psutil
import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score
from sklearn.ensemble import VotingClassifier, StackingClassifier, RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression

os.makedirs("../models", exist_ok=True)
os.makedirs("../results", exist_ok=True)

# 1. Load the NEWLY trained base models
models_dir = "../models"
rf = joblib.load(os.path.join(models_dir, "Random_Forest.pkl"))
ada = joblib.load(os.path.join(models_dir, "AdaBoost.pkl"))
gb = joblib.load(os.path.join(models_dir, "Gradient_Boosting.pkl"))
knn = joblib.load(os.path.join(models_dir, "KNN.pkl"))
nb = joblib.load(os.path.join(models_dir, "Naive_Bayes.pkl"))

# 2. Load the 145-feature data
data_dir = "../dataset/processed"
X_train = np.load(os.path.join(data_dir, "X_train.npy"))
y_train = np.load(os.path.join(data_dir, "y_train.npy"))
X_test = np.load(os.path.join(data_dir, "X_test.npy"))
y_test = np.load(os.path.join(data_dir, "y_test.npy"))

# 3. Define ALL 8 Ensemble Models
ensembles = {
    "Hard_Voting": VotingClassifier(estimators=[('rf', rf), ('gb', gb), ('knn', knn)], voting='hard'),
    "Soft_Voting": VotingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada)], voting='soft'),
    "Stacking_Top3_LR": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada)], final_estimator=LogisticRegression(max_iter=1000)),
    "Stacking_Top3_GB": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada)], final_estimator=GradientBoostingClassifier(random_state=42)),
    "Stacking_Top3_RF": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada)], final_estimator=RandomForestClassifier(random_state=42, n_jobs=-1)),
    "Stacking_All_LR": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada), ('knn', knn), ('nb', nb)], final_estimator=LogisticRegression(max_iter=1000)),
    "Stacking_All_GB": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada), ('knn', knn), ('nb', nb)], final_estimator=GradientBoostingClassifier(random_state=42)),
    "Stacking_All_RF": StackingClassifier(estimators=[('rf', rf), ('gb', gb), ('ada', ada), ('knn', knn), ('nb', nb)], final_estimator=RandomForestClassifier(random_state=42, n_jobs=-1))
}

ensemble_results = []

for name, model in ensembles.items():
    print("=" * 70)
    print(f"Training {name}...")

    process = psutil.Process(os.getpid())
    cpu_before = psutil.cpu_percent(interval=None)
    ram_before = process.memory_info().rss

    start_train = time.time()
    model.fit(X_train, y_train)
    end_train = time.time()

    cpu_after = psutil.cpu_percent(interval=None)
    ram_after = process.memory_info().rss

    start_predict = time.time()
    predictions = model.predict(X_test)
    end_predict = time.time()

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions)
    recall = recall_score(y_test, predictions)
    f1 = f1_score(y_test, predictions)

    tn, fp, fn, tp = confusion_matrix(y_test, predictions, labels=[0, 1]).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0

    roc = np.nan
    if hasattr(model, "predict_proba"):
        try:
            probabilities = model.predict_proba(X_test)[:, 1]
            roc = roc_auc_score(y_test, probabilities)
        except Exception:
            pass

    filename = f"../models/{name}.pkl"
    joblib.dump(model, filename)
    model_size = os.path.getsize(filename) / (1024 * 1024)

    ensemble_results.append({
        "Model": name,
        "Features": X_train.shape[1], # This will now correctly be 145
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "False Positive Rate": fpr,
        "ROC-AUC": roc,
        "CPU Usage (%)": (cpu_before + cpu_after) / 2,
        "RAM Usage (MB)": abs((ram_after - ram_before) / (1024 * 1024)), 
        "Training Time (s)": end_train - start_train,
        "Prediction Time (s)": end_predict - start_predict,
        "Model Size (MB)": model_size
    })

    print(f"✓ {name} completed. | Specificity: {specificity:.4f} | FPR: {fpr:.4f} | Sum: {specificity + fpr:.4f}")

ensemble_df = pd.DataFrame(ensemble_results)
ensemble_df.to_csv("../results/ensemble_models_results.csv", index=False)
print("\n✓ All 8 ensemble models trained and saved!")
```

```python
base_df = pd.read_csv("../results/base_models_results.csv")
ensemble_df = pd.read_csv("../results/ensemble_models_results.csv")

final_df = pd.concat([base_df, ensemble_df], ignore_index=True)

# TABLE 1: PERFORMANCE
perf_metrics = ["Accuracy", "Precision", "Recall", "F1", "Specificity", "ROC-AUC"]
perf_df = final_df[["Model"] + perf_metrics].copy()
perf_df["False Positive Rate"] = final_df["False Positive Rate"]

ranked_perf = perf_df[perf_metrics].rank(ascending=False, method='average')
ranked_perf["False Positive Rate"] = perf_df["False Positive Rate"].rank(ascending=True, method='average')

perf_df["Average Performance Rank"] = ranked_perf.mean(axis=1).round(2)
perf_df = perf_df.sort_values("Average Performance Rank").reset_index(drop=True)
perf_df.insert(0, "Overall Rank", range(1, len(perf_df) + 1))
perf_df[perf_metrics + ["False Positive Rate"]] = perf_df[perf_metrics + ["False Positive Rate"]].round(4)

print("\n" + "="*80)
print("TABLE 1: PERFORMANCE EVALUATION (All 13 Models)")
print("="*80)
display(perf_df)

# TABLE 2: EFFICIENCY 
eff_metrics = ["Prediction Time (s)", "Model Size (MB)"]

eff_df = final_df[["Model"] + eff_metrics].copy()
ranked_eff = eff_df[eff_metrics].rank(ascending=True, method='average')

eff_df["Average Efficiency Rank"] = ranked_eff.mean(axis=1).round(2)
eff_df = eff_df.sort_values("Average Efficiency Rank").reset_index(drop=True)
eff_df.insert(0, "Overall Rank", range(1, len(eff_df) + 1))
eff_df[eff_metrics] = eff_df[eff_metrics].round(4)

print("\n" + "="*80)
print("TABLE 2: EFFICIENCY EVALUATION (All 13 Models)")
print("="*80)
display(eff_df)


perf_df.to_csv("../results/final_table1_performance.csv", index=False)
eff_df.to_csv("../results/final_table2_efficiency.csv", index=False)
print("\n✓ Final unified tables saved to ../results/!")
```
